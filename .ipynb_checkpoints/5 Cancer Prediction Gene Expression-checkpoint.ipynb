```python
import numpy as np
import pandas as pd

df = pd.read_csv("cancer_gene_expression.csv")
```

```python

X = df.iloc[:, :-1]

y = df.iloc[:, -1]
```

```python

X
```

```python

from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

y_encoded = le.fit_transform(y)
```

```python

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split( X, y_encoded, test_size=0.2, random_state=42,stratify=y_encoded)
```

```python
# Pipeline

from sklearn.pipeline import Pipeline

from sklearn.preprocessing import MinMaxScaler

from sklearn.feature_selection import SelectKBest, mutual_info_classif

from sklearn.multiclass import OneVsRestClassifier

from sklearn.ensemble import RandomForestClassifier
```

```python

pipeline = Pipeline([
    ("scaler", MinMaxScaler()),

    ("feature_selection", SelectKBest(score_func=mutual_info_classif, k=250)),

    ("model", OneVsRestClassifier(
        RandomForestClassifier(n_estimators=200, max_depth=None, max_features=0.2, class_weight="balanced",random_state=42, n_jobs=-1)
    ))
])
```

```python

pipeline
```

```python

pipeline.fit(X_train, y_train)
```

```python

y_pred = pipeline.predict(X_test)

y_prob = pipeline.predict_proba(X_test)
```

```python

from sklearn.metrics import classification_report, confusion_matrix, balanced_accuracy_score

print("Test Balanced Accuracy:", balanced_accuracy_score(y_test, y_pred))

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred, target_names=le.classes_))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred))
```

```python

from sklearn.metrics import roc_auc_score
from sklearn.preprocessing import label_binarize

# Binarize the test labels
y_test_bin = label_binarize(y_test, classes=np.unique(y_encoded))

# Compute ROC-AUC
roc_auc = roc_auc_score(
    y_test_bin,
    y_prob,
    multi_class="ovr",
    average="weighted"   # or "macro"
)

print("\nMulti-class ROC-AUC (OvR):", roc_auc)
```
