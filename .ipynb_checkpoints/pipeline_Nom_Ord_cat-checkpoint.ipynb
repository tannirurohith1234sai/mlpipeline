```python
import numpy as np
import pandas as pd

df = pd.read_csv("breast-cancer.csv")

df.head()
```

```python

print("Shape:", df.shape)
```

```python

print(df.columns)
```

```python

df.info()
```

```python
# Remove single quotes

df = df.map(
    lambda x: x.replace("'", "") if isinstance(x, str) else x)
```

```python
# Replace ? with NaN

df = df.replace("?", np.nan)
```

```python
# Check missing values

df.isnull().sum()
```

```python

X = df.drop(columns=["Class"])

print("X shape:", X.shape)
```

```python

y = df["Class"]
print("y shape:", y.shape)
```

```python

ordinal_orders = {
    "age": ["20-29", "30-39", "40-49", "50-59", "60-69", "70-79"],

    "tumor-size": [
        "0-4", "5-9", "10-14", "15-19", "20-24",
        "25-29", "30-34", "35-39", "40-44", "45-49", "50-54"
    ],

    "inv-nodes": [
        "0-2", "3-5", "6-8", "9-11", "12-14",
        "15-17", "18-20", "21-23", "24-26"
    ],

    "deg-malig": ["1", "2", "3"]
}

ordinal_cols = list(ordinal_orders.keys())

print("Ordinal:", ordinal_cols)
```

```python

nominal_cols = [
    "menopause", "node-caps", "breast",
    "breast-quad", "irradiat"
]

print("Nominal:", nominal_cols)
```

```python

from sklearn.model_selection import train_test_split

from sklearn.preprocessing import LabelEncoder

target_encoder = LabelEncoder()
y_encoded = target_encoder.fit_transform(y)

print("Classes:", target_encoder.classes_)
```

```python

X_train, X_test, y_train, y_test = train_test_split( X, y_encoded, test_size=0.20,stratify=y_encoded)

print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
```

```python

from sklearn.pipeline import Pipeline

from sklearn.impute import SimpleImputer

from sklearn.preprocessing import OrdinalEncoder, OneHotEncoder

from sklearn.compose import ColumnTransformer
```

```python

ordinal_categories = [
    ordinal_orders[col] for col in ordinal_cols
]
ordinal_categories
```

```python

ordinal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),

    ("encoder", OrdinalEncoder( categories=ordinal_categories, handle_unknown="use_encoded_value", unknown_value=-1))
])
```

```python

nominal_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),

    ("encoder", OneHotEncoder(
        handle_unknown="ignore",
        sparse_output=False
    ))
])
```

```python

preprocessor = ColumnTransformer([
    ("ordinal", ordinal_pipeline, ordinal_cols),
    ("nominal", nominal_pipeline, nominal_cols)
])
```

```python
preprocessor
```

```python

from sklearn.naive_bayes import CategoricalNB

nb_pipeline = Pipeline([
    ("preprocessor", preprocessor),

    ("classifier", CategoricalNB(alpha=1.0))
])
```

```python

nb_pipeline.fit(X_train, y_train)
```

```python

y_pred = nb_pipeline.predict(X_test)

y_pred[:5]
```

```python

from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

accuracy = accuracy_score(y_test, y_pred)

print(f"Accuracy: {accuracy:.2f}")
```

```python

cm = confusion_matrix(y_test, y_pred)

cm_df = pd.DataFrame(
    cm,
    index=target_encoder.classes_,
    columns=target_encoder.classes_
)

cm_df
```

```python

print(classification_report( y_test, y_pred, target_names=target_encoder.classes_, zero_division=0
))
```

```python

y_prob = nb_pipeline.predict_proba(X_test)

y_prob[:5]
```

```python

result = pd.DataFrame({
    "Actual": target_encoder.inverse_transform(y_test),
    "Predicted": target_encoder.inverse_transform(y_pred)
})

result.head(10)
```
